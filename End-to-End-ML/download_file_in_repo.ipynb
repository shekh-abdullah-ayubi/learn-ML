```python
import urllib.request
import os
import pandas as pd

url=['https://raw.githubusercontent.com/nyandwi/public_datasets/master/housing.csv',
    'https://raw.githubusercontent.com/nyandwi/public_datasets/master/mobile_price_train.csv',
    'https://raw.githubusercontent.com/nyandwi/public_datasets/master/mobile_price_test.csv',
    ]
repo_dir='/workspaces/learn-ML/End-to-End-ML/kaggle_datasets'
for i in url:
    filename=os.path.basename(i)
    os.makedirs(repo_dir,exist_ok=True)
    print(filename)
    file_path=os.path.join(repo_dir,filename)
    data_path =  urllib.request.urlretrieve(i)[0]
    data = pd.read_csv(i)
    data.to_csv(file_path,index=False)
```
